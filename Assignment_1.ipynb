## Q3 – Strings
Accept a string. Count characters, vowels, consonants, reverse it, and check if it is a palindrome.

```python
# Q3: String Operations

s = "racecar"
print("String:", s)

char_count = len(s)
print("Total characters:", char_count)

vowels = "aeiouAEIOU"
vowel_count = 0
consonant_count = 0

for ch in s:
    if ch.isalpha():
        if ch in vowels:
            vowel_count += 1
        else:
            consonant_count += 1

print("Vowels:", vowel_count)
print("Consonants:", consonant_count)

reversed_s = s[::-1]
print("Reversed string:", reversed_s)

if s == reversed_s:
    print(f"\"{s}\" is a palindrome")
else:
    print(f"\"{s}\" is not a palindrome")
```

## Q5 – GPA Calculator
Calculate GPA using grade points and credits. GPA = Σ(GP × Credit) / Σ(Credits).

```python
# Q5: GPA Calculator

subjects = [
    ("Mathematics", 9, 4),
    ("Physics", 8, 3),
    ("Chemistry", 7, 3),
    ("English", 8, 2),
    ("Computer Science", 10, 4)
]

print("Subject-wise Details:")
print(f"{'Subject':<20} {'Grade Point':<12} {'Credits'}")

total_weighted = 0
total_credits = 0

for subject, gp, credit in subjects:
    print(f"{subject:<20} {gp:<12} {credit}")
    total_weighted += gp * credit
    total_credits += credit

print(f"Total Weighted Points: {total_weighted}")
print(f"Total Credits: {total_credits}")

gpa = total_weighted / total_credits
print(f"\nGPA = {total_weighted} / {total_credits} = {gpa:.2f}")
```

## Q6 – Functions and Iteration
Write functions for factorial, Fibonacci series, and prime number check.

```python
# Q6: Functions and Iteration

def factorial(n):
    """Calculate factorial of n"""
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

def fibonacci(n):
    """Generate first n Fibonacci numbers"""
    fib_series = []
    a, b = 0, 1
    for _ in range(n):
        fib_series.append(a)
        a, b = b, a + b
    return fib_series

def is_prime(n):
    """Check if n is a prime number"""
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print("Factorial of 5:", factorial(5))
print("Factorial of 7:", factorial(7))
print()

print("First 10 Fibonacci numbers:", fibonacci(10))
print()

test_numbers = [2, 7, 10, 13, 20, 29]
for num in test_numbers:
    if is_prime(num):
        print(f"{num} is a Prime number")
    else:
        print(f"{num} is not a Prime number")
```

## Q8 – Exception Handling
Perform division with try-except. Handle both normal and division-by-zero cases.

```python
# Q8: Exception Handling

def divide(a, b):
    """Divide a by b with exception handling"""
    try:
        result = a / b
        print(f"{a} / {b} = {result}")
    except ZeroDivisionError:
        print(f"Error: Cannot divide {a} by zero!")
    except TypeError:
        print("Error: Invalid input type!")

print("Case 1: Normal Division")
divide(10, 3)

print()

print("Case 2: Division by Zero")
divide(10, 0)

print()

print("Case 3: Another Normal Division")
divide(100, 4)
```

## Q10 – Mini Programming Problem: Student Record System
Menu-driven student record system using dictionary. Demonstrate add, update, search, delete, and display operations.

```python
# Q10: Student Record System using Dictionary

records = {}

def add_student(roll, name, marks):
    """Add a new student record"""
    if roll in records:
        print(f"  Student with Roll No {roll} already exists!")
    else:
        records[roll] = {"name": name, "marks": marks}
        print(f"  Added: Roll={roll}, Name={name}, Marks={marks}")

def update_student(roll, marks):
    """Update marks of an existing student"""
    if roll in records:
        old_marks = records[roll]["marks"]
        records[roll]["marks"] = marks
        print(f"  Updated Roll {roll}: Marks {old_marks} -> {marks}")
    else:
        print(f"  Student with Roll No {roll} not found!")

def search_student(roll):
    """Search for a student by roll number"""
    if roll in records:
        s = records[roll]
        print(f"  Found: Roll={roll}, Name={s['name']}, Marks={s['marks']}")
    else:
        print(f"  Student with Roll No {roll} not found!")

def delete_student(roll):
    """Delete a student record"""
    if roll in records:
        removed = records.pop(roll)
        print(f"  Deleted: Roll={roll}, Name={removed['name']}")
    else:
        print(f"  Student with Roll No {roll} not found!")

def display_all():
    """Display all student records"""
    if not records:
        print("  No records found!")
        return
    print(f"  {'Roll':<8} {'Name':<15} {'Marks'}")
    print("  " )
    for roll, info in records.items():
        print(f"  {roll:<8} {info['name']:<15} {info['marks']}")


print("1. Adding Students")
add_student(101, "Alice", 85)
add_student(102, "Bob", 72)
add_student(103, "Charlie", 90)
add_student(104, "Diana", 68)

print("\n2. Display All Records")
display_all()

print("\n3. Search for a Student")
search_student(102)
search_student(999)

print("\n4. Update Student Marks")
update_student(102, 88)

print("\n5. Display After Update")
display_all()

print("\n6. Delete a Student")
delete_student(104)

print("\n7. Display After Deletion")

display_all()
```
